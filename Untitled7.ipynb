```python
#task 02
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Load dataset
input_file = Path("train.csv")
output_dir = Path("titanic_eda_outputs")
output_dir.mkdir(exist_ok=True)

df = pd.read_csv(input_file)

# -----------------------------
# DATA CLEANING
# -----------------------------
df = df.drop_duplicates().copy()

# Age: impute with median within Pclass + Sex groups
df["Age"] = df["Age"].fillna(
    df.groupby(["Pclass", "Sex"])["Age"].transform("median")
)

# Embarked: fill missing values with the mode
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

# Cabin: create a Deck feature because Cabin has many missing values
df["Deck"] = df["Cabin"].str[0].fillna("Unknown")
df["CabinKnown"] = df["Cabin"].notna().astype(int)

# -----------------------------
# FEATURE ENGINEERING
# -----------------------------
df["FamilySize"] = df["SibSp"] + df["Parch"] + 1
df["IsAlone"] = (df["FamilySize"] == 1).astype(int)

df["Title"] = (
    df["Name"]
    .str.extract(r",\s*([^.]*)\.", expand=False)
    .str.strip()
)
common_titles = ["Mr", "Miss", "Mrs", "Master"]
df["Title"] = df["Title"].where(df["Title"].isin(common_titles), "Rare")

# Save cleaned dataset
df.to_csv(output_dir / "titanic_cleaned.csv", index=False)

# -----------------------------
# SUMMARY STATISTICS
# -----------------------------
print("Rows:", len(df))
print("Columns:", df.shape[1])
print(f"Overall survival rate: {df['Survived'].mean()*100:.1f}%")
print("\nMissing values after cleaning:")
print(df.isna().sum())

print("\nSurvival by sex (%):")
print((df.groupby("Sex")["Survived"].mean()*100).round(1))

print("\nSurvival by passenger class (%):")
print((df.groupby("Pclass")["Survived"].mean()*100).round(1))

print("\nSurvival by embarkation port (%):")
print((df.groupby("Embarked")["Survived"].mean()*100).round(1))

# -----------------------------
# EDA DASHBOARD
# -----------------------------
fig, axes = plt.subplots(3, 2, figsize=(15, 18))
fig.suptitle(
    "Titanic Dataset — Data Cleaning & Exploratory Data Analysis",
    fontsize=20, fontweight="bold", y=0.995
)

# 1. Survival by sex
sex_survival = df.groupby("Sex")["Survived"].mean() * 100
ax = axes[0, 0]
bars = ax.bar(sex_survival.index, sex_survival.values)
ax.set_title("Survival Rate by Sex", fontweight="bold")
ax.set_ylabel("Survival rate (%)")
ax.set_ylim(0, 100)
ax.grid(axis="y", alpha=0.25)
for bar, value in zip(bars, sex_survival.values):
    ax.text(bar.get_x()+bar.get_width()/2, value+2,
            f"{value:.1f}%", ha="center", fontweight="bold")

# 2. Survival by class
class_survival = df.groupby("Pclass")["Survived"].mean() * 100
ax = axes[0, 1]
bars = ax.bar(class_survival.index.astype(str), class_survival.values)
ax.set_title("Survival Rate by Passenger Class", fontweight="bold")
ax.set_xlabel("Passenger class")
ax.set_ylabel("Survival rate (%)")
ax.set_ylim(0, 100)
ax.grid(axis="y", alpha=0.25)
for bar, value in zip(bars, class_survival.values):
    ax.text(bar.get_x()+bar.get_width()/2, value+2,
            f"{value:.1f}%", ha="center", fontweight="bold")

# 3. Age distribution
ax = axes[1, 0]
ax.hist(df.loc[df["Survived"] == 0, "Age"], bins=25, alpha=0.65,
        label="Did not survive")
ax.hist(df.loc[df["Survived"] == 1, "Age"], bins=25, alpha=0.65,
        label="Survived")
ax.set_title("Age Distribution by Survival", fontweight="bold")
ax.set_xlabel("Age")
ax.set_ylabel("Number of passengers")
ax.legend()
ax.grid(axis="y", alpha=0.2)

# 4. Fare distribution
ax = axes[1, 1]
ax.hist(df.loc[df["Survived"] == 0, "Fare"], bins=30, alpha=0.65,
        label="Did not survive")
ax.hist(df.loc[df["Survived"] == 1, "Fare"], bins=30, alpha=0.65,
        label="Survived")
ax.set_title("Fare Distribution by Survival", fontweight="bold")
ax.set_xlabel("Fare")
ax.set_ylabel("Number of passengers")
ax.legend()
ax.grid(axis="y", alpha=0.2)

# 5. Family size vs survival
family_survival = df.groupby("FamilySize")["Survived"].mean() * 100
family_survival = family_survival[family_survival.index <= 8]
ax = axes[2, 0]
ax.plot(family_survival.index, family_survival.values,
        marker="o", linewidth=2)
ax.set_title("Survival Rate by Family Size", fontweight="bold")
ax.set_xlabel("Family size")
ax.set_ylabel("Survival rate (%)")
ax.set_xticks(family_survival.index)
ax.set_ylim(0, 100)
ax.grid(alpha=0.25)

# 6. Class x Sex heatmap
heat = df.pivot_table(
    index="Pclass", columns="Sex",
    values="Survived", aggfunc="mean"
) * 100
ax = axes[2, 1]
im = ax.imshow(heat.values, aspect="auto")
ax.set_title("Survival Rate: Class × Sex", fontweight="bold")
ax.set_xticks(range(len(heat.columns)))
ax.set_xticklabels(heat.columns)
ax.set_yticks(range(len(heat.index)))
ax.set_yticklabels([f"Class {x}" for x in heat.index])

for i in range(heat.shape[0]):
    for j in range(heat.shape[1]):
        ax.text(j, i, f"{heat.iloc[i,j]:.1f}%",
                ha="center", va="center", fontweight="bold")

fig.colorbar(im, ax=ax, fraction=0.046, pad=0.04,
             label="Survival rate (%)")

plt.tight_layout(rect=[0, 0, 1, 0.98])
plt.savefig(
    output_dir / "titanic_eda_dashboard.png",
    dpi=180, bbox_inches="tight"
)
plt.show()
```
